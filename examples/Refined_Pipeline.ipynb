# AutoMorphoTrack 2.5.0 particle example
Run from the release repository directory after installing the pinned wheel. The adapter checks both numerical namespaces. The input is a synthetic fixture with known motion.

```python
from pathlib import Path
import importlib.util
import json
import sys
from importlib import metadata
root = Path.cwd()
if root.name == 'examples':
    root = root.parent
assert (root / 'validation').is_dir(), 'Open this notebook from the release repository or its examples directory'
assert metadata.version('automorphotrack') == '2.5.0'
spec = importlib.util.spec_from_file_location('verified_adapter', root / 'examples/refined_verified.py')
adapter = importlib.util.module_from_spec(spec)
spec.loader.exec_module(adapter)
print(sys.executable)
print(metadata.version('automorphotrack'))
```

```python
import os
output = Path(os.environ.get('AMT_NOTEBOOK_OUTPUT', 'refined_notebook_output'))
result = adapter.run_verified(root / 'validation/refined_motion.ome.tif', output, root / 'validation/refined_motion_config.json', root / 'validation')
print(json.dumps({'source_sha256': result['manifest']['source_sha256'], 'output': str(output)}, indent=2))
```
